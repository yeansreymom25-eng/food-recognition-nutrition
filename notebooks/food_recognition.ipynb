# 6. Train, Validation and Test Split

This section divides the dataset into training, validation, and testing sets before model training.

```python
import os
import random
import shutil

random.seed(42)

source = "/content/gdrive/MyDrive/Food_Recognition_Project/dataset"
output = "/content/gdrive/MyDrive/Food_Recognition_Project/split_data"

for split in ["train", "validation", "test"]:
    os.makedirs(os.path.join(output, split), exist_ok=True)

classes = [
    folder for folder in os.listdir(source)
    if os.path.isdir(os.path.join(source, folder))
]

for class_name in classes:

    class_path = os.path.join(source, class_name)

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    total = len(images)

    train_end = int(total * 0.70)
    validation_end = train_end + int(total * 0.20)

    train_images = images[:train_end]
    validation_images = images[train_end:validation_end]
    test_images = images[validation_end:]

    for split_name, image_list in [
        ("train", train_images),
        ("validation", validation_images),
        ("test", test_images)
    ]:

        destination = os.path.join(
            output,
            split_name,
            class_name
        )

        os.makedirs(destination, exist_ok=True)

        for image in image_list:
            shutil.copy2(
                os.path.join(class_path, image),
                os.path.join(destination, image)
            )

    print(
        class_name,
        "| Train:", len(train_images),
        "| Validation:", len(validation_images),
        "| Test:", len(test_images)
    )

print("\nDataset split finished!")
```

7.1 Image Preprocessing and Data Augmentation

This section prepares the images before training. All images are resized to 224×224 pixels, converted to tensors, and normalized. Training images also use simple data augmentation such as horizontal flipping and small rotation to help the model learn better.

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
from torchvision import transforms

# Training images
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Validation and testing images
val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Image transformations ready!")
```

What this code does
Resize(224,224) → makes every image the correct size for MobileNetV2
RandomHorizontalFlip() → sometimes flips training images
RandomRotation(10) → slightly rotates training images
ToTensor() → changes the image into data PyTorch can use
Normalize() → prepares the image values for the pretrained MobileNetV2

7.2 Load Train, Validation, and Test Datasets

This section loads the training, validation, and testing image folders using PyTorch ImageFolder. The folder names are automatically used as the class labels for the 20 food classes.

```python
import os
from torchvision import datasets

data_path = "/content/gdrive/MyDrive/Food_Recognition_Project/split_data"

train_dataset = datasets.ImageFolder(
    os.path.join(data_path, "train"),
    transform=train_transform
)

validation_dataset = datasets.ImageFolder(
    os.path.join(data_path, "validation"),
    transform=val_test_transform
)

test_dataset = datasets.ImageFolder(
    os.path.join(data_path, "test"),
    transform=val_test_transform
)

print("Train images:", len(train_dataset))
print("Validation images:", len(validation_dataset))
print("Test images:", len(test_dataset))

print("\nNumber of classes:", len(train_dataset.classes))
print(train_dataset.classes)
```

7.3 Create PyTorch DataLoaders

This section loads the dataset in small batches so the model can train efficiently without loading all images into memory at once.

```python
from torch.utils.data import DataLoader

batch_size = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=2
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=2
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=2
)

print("DataLoaders ready!")
print("Batch size:", batch_size)
```

7.4 Load and Prepare MobileNetV2

This section loads a pretrained MobileNetV2 model, freezes the pretrained feature layers, and changes the final classifier to predict my 20 food classes.

```python
import torch.nn as nn
from torchvision import models
from torchvision.models import MobileNet_V2_Weights

# Load pretrained MobileNetV2
weights = MobileNet_V2_Weights.DEFAULT
model = models.mobilenet_v2(weights=weights)

print("MobileNetV2 loaded!")
```

```python
for param in model.features.parameters():
    param.requires_grad = False

print("Pretrained feature layers frozen!")
```

Why?

MobileNetV2 already learned useful image features from ImageNet, so we keep those features and train mainly the new food classifier.

```python
num_classes = 20

model.classifier[1] = nn.Linear(
    model.last_channel,
    num_classes
)

model = model.to(device)

print("Model ready for", num_classes, "food classes!")
```

```python
import os
import torch

model_dir = "/content/gdrive/MyDrive/Food_Recognition_Project/model"
os.makedirs(model_dir, exist_ok=True)

# Continue from Epoch 2 to Epoch 5
for epoch in range(1, num_epochs):

    # -------------------
    # TRAINING
    # -------------------
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_loss = running_loss / len(train_loader)
    train_accuracy = 100 * correct / total

    # -------------------
    # VALIDATION
    # -------------------
    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in validation_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()

    val_loss = val_running_loss / len(validation_loader)
    val_accuracy = 100 * val_correct / val_total

    # Save results
    train_losses.append(train_loss)
    train_accuracies.append(train_accuracy)

    val_losses.append(val_loss)
    val_accuracies.append(val_accuracy)

    print(f"Epoch {epoch+1}/{num_epochs}")
    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.2f}%"
    )
    print(
        f"Validation Loss: {val_loss:.4f} | "
        f"Validation Accuracy: {val_accuracy:.2f}%"
    )

    # Save the best model
    if val_accuracy > best_val_accuracy:

        best_val_accuracy = val_accuracy

        torch.save(
            model.state_dict(),
            os.path.join(model_dir, "best_mobilenetv2.pth")
        )

        print("✅ New best model saved!")

    print("-" * 50)

print("\nTraining finished!")
print(f"Best Validation Accuracy: {best_val_accuracy:.2f}%")
```

11. Final Food Prediction Function

This section combines the trained MobileNetV2 model with the nutrition database. The function accepts a food image and portion size, predicts the food class and confidence score, and displays the estimated nutritional information.

```python
import torch
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms

# Class names must follow the same order used during training
class_names = [
    'almond', 'apple', 'banana', 'broccoli', 'carrot',
    'cashew', 'chocolate_cake', 'cucumber', 'cupcake',
    'donut', 'french_fries', 'hamburger', 'hot_dog',
    'ice_cream', 'orange', 'peanut', 'pizza',
    'strawberry', 'tomato', 'walnut'
]

# Load nutrition database
nutrition_path = "/content/gdrive/MyDrive/Food_Recognition_Project/nutrition/food_nutrition.csv"

nutrition_df = pd.read_csv(nutrition_path)
nutrition_dict = nutrition_df.set_index("food").to_dict("index")

# Transformation for new images
prediction_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


def predict_food(image_path, portion_grams=100):

    # Load image
    original_image = Image.open(image_path).convert("RGB")

    # Prepare image
    image_tensor = prediction_transform(original_image)
    image_tensor = image_tensor.unsqueeze(0).to(device)

    # Prediction
    model.eval()

    with torch.no_grad():

        output = model(image_tensor)

        probabilities = torch.softmax(output, dim=1)

        confidence, predicted_label = torch.max(
            probabilities, 1
        )

    predicted_food = class_names[predicted_label.item()]
    confidence_percent = confidence.item() * 100

    # Nutrition information
    nutrition = nutrition_dict[predicted_food]

    ratio = portion_grams / 100

    calories = nutrition["calories_kcal"] * ratio
    protein = nutrition["protein_g"] * ratio
    carbs = nutrition["carbs_g"] * ratio
    fat = nutrition["fat_g"] * ratio

    # Display result
    print("=" * 35)
    print("FOOD RECOGNITION RESULT")
    print("=" * 35)

    print(
        "Food:",
        predicted_food.replace("_", " ").title()
    )

    print(
        f"Confidence: {confidence_percent:.2f}%"
    )

    print(
        "Category:",
        nutrition["category"]
    )

    print("\nEstimated Nutrition")
    print("----------------------")

    print(f"Portion Size: {portion_grams} g")
    print(f"Calories: {calories:.2f} kcal")
    print(f"Protein: {protein:.2f} g")
    print(f"Carbohydrates: {carbs:.2f} g")
    print(f"Fat: {fat:.2f} g")

    # Display image
    plt.figure(figsize=(5, 5))

    plt.imshow(original_image)

    plt.title(
        f"{predicted_food.replace('_', ' ').title()}\n"
        f"Confidence: {confidence_percent:.2f}%"
    )

    plt.axis("off")
    plt.show()
```

12. Fine-Tuning MobileNetV2

This section fine-tunes the last layers of the pretrained MobileNetV2 model. Instead of keeping all feature layers frozen, I allow the last few layers to learn from my food dataset using a smaller learning rate.

```python
# Freeze all feature layers first
for param in model.features.parameters():
    param.requires_grad = False

# Unfreeze the last 3 feature blocks
for layer in model.features[-3:]:
    for param in layer.parameters():
        param.requires_grad = True

print("Last 3 MobileNetV2 feature blocks are ready for fine-tuning.")
```

```python
import torch.optim as optim

optimizer = optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=0.0001
)

print("Fine-tuning optimizer ready!")
print("Learning rate: 0.0001")
```

Why use 0.0001 now?
Before, we used:0.001
For fine-tuning we use a smaller learning rate:0.0001
because the pretrained layers already contain useful knowledge. We only want to adjust them a little, not change them too much.

12.1 Fine-Tune the Model

This section trains the last few MobileNetV2 feature layers for 3 additional epochs using a smaller learning rate. The validation accuracy is checked after each epoch, and the best fine-tuned model is saved separately.

```python
fine_tune_epochs = 3

best_fine_val_accuracy = 0.0

fine_train_accuracies = []
fine_val_accuracies = []

for epoch in range(fine_tune_epochs):

    # -------------------
    # TRAINING
    # -------------------
    model.train()

    correct = 0
    total = 0
    running_loss = 0.0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_loss = running_loss / len(train_loader)
    train_accuracy = 100 * correct / total

    # -------------------
    # VALIDATION
    # -------------------
    model.eval()

    val_correct = 0
    val_total = 0
    val_running_loss = 0.0

    with torch.no_grad():

        for images, labels in validation_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()

    val_loss = val_running_loss / len(validation_loader)
    val_accuracy = 100 * val_correct / val_total

    fine_train_accuracies.append(train_accuracy)
    fine_val_accuracies.append(val_accuracy)

    print(f"Fine-Tune Epoch {epoch+1}/{fine_tune_epochs}")
    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.2f}%"
    )
    print(
        f"Validation Loss: {val_loss:.4f} | "
        f"Validation Accuracy: {val_accuracy:.2f}%"
    )

    # Save best fine-tuned model
    if val_accuracy > best_fine_val_accuracy:

        best_fine_val_accuracy = val_accuracy

        torch.save(
            model.state_dict(),
            "/content/drive/MyDrive/Food_Recognition_Project/model/best_finetuned_mobilenetv2.pth"
        )

        print("✅ Best fine-tuned model saved!")

    print("-" * 50)

print("\nFine-tuning finished!")
print(f"Best Fine-Tuned Validation Accuracy: {best_fine_val_accuracy:.2f}%")
```

14.2 Confidence Threshold for Uncertain Predictions

This section adds a confidence threshold to make the system more reliable. If the model confidence is below 50%, the system shows "Uncertain Prediction" and does not display category or nutrition information.

```python
def predict_food(image_path, portion_grams=100):

    # Load image
    original_image = Image.open(image_path).convert("RGB")

    # Prepare image
    image_tensor = prediction_transform(original_image)
    image_tensor = image_tensor.unsqueeze(0).to(device)

    # Prediction
    model.eval()

    with torch.no_grad():

        output = model(image_tensor)

        probabilities = torch.softmax(output, dim=1)

        confidence, predicted_label = torch.max(
            probabilities, 1
        )

    predicted_food = class_names[predicted_label.item()]
    confidence_percent = confidence.item() * 100

    print("=" * 35)
    print("FOOD RECOGNITION RESULT")
    print("=" * 35)

    # ----------------------------------
    # LOW CONFIDENCE
    # ----------------------------------
    if confidence_percent < 50:

        print("Result: Uncertain Prediction")
        print(
            "Top Prediction:",
            predicted_food.replace("_", " ").title()
        )
        print(f"Confidence: {confidence_percent:.2f}%")

        print("\nThe confidence is too low.")
        print("Category and nutrition are not shown.")

        plt.figure(figsize=(5, 5))
        plt.imshow(original_image)

        plt.title(
            f"Uncertain Prediction\n"
            f"Confidence: {confidence_percent:.2f}%"
        )

        plt.axis("off")
        plt.show()

        return

    # ----------------------------------
    # CONFIDENT PREDICTION
    # ----------------------------------
    nutrition = nutrition_dict[predicted_food]

    ratio = portion_grams / 100

    calories = nutrition["calories_kcal"] * ratio
    protein = nutrition["protein_g"] * ratio
    carbs = nutrition["carbs_g"] * ratio
    fat = nutrition["fat_g"] * ratio

    print(
        "Food:",
        predicted_food.replace("_", " ").title()
    )

    print(f"Confidence: {confidence_percent:.2f}%")
    print("Category:", nutrition["category"])

    print("\nEstimated Nutrition")
    print("----------------------")
    print(f"Portion Size: {portion_grams} g")
    print(f"Calories: {calories:.2f} kcal")
    print(f"Protein: {protein:.2f} g")
    print(f"Carbohydrates: {carbs:.2f} g")
    print(f"Fat: {fat:.2f} g")

    plt.figure(figsize=(5, 5))
    plt.imshow(original_image)

    plt.title(
        f"{predicted_food.replace('_', ' ').title()}\n"
        f"Confidence: {confidence_percent:.2f}%"
    )

    plt.axis("off")
    plt.show()
```

15.3 Load Final Model for Further Fine-Tuning

This section loads the previous best fine-tuned MobileNetV2 model. The last three feature blocks are unfrozen again so the model can learn from the newly added diverse training images.

```python
import torch
import torch.nn as nn
from torchvision import models

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Create MobileNetV2 structure
model = models.mobilenet_v2(weights=None)

# Change classifier to 20 food classes
model.classifier[1] = nn.Linear(
    model.last_channel,
    20
)

# Load previous best model
model_path = (
    "/content/drive_new/MyDrive/"
    "Food_Recognition_Project/model/"
    "best_finetuned_mobilenetv2.pth"
)

model.load_state_dict(
    torch.load(model_path, map_location=device)
)

model = model.to(device)

# Freeze all feature layers
for param in model.features.parameters():
    param.requires_grad = False

# Unfreeze last 3 feature blocks
for layer in model.features[-3:]:
    for param in layer.parameters():
        param.requires_grad = True

print("Previous best fine-tuned model loaded!")
print("Device:", device)
print("Ready for improvement training!")
```

15.4 Further Fine-Tuning with Improvement Data

This section further fine-tunes the previous best MobileNetV2 model using the updated training dataset. The additional diverse images help the model learn difficult real-world food appearances. A small learning rate is used to avoid changing the pretrained knowledge too much.

```python
import torch
import torch.nn as nn
import torch.optim as optim
import os

# Loss function
criterion = nn.CrossEntropyLoss()

# Smaller learning rate for further fine-tuning
optimizer = optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=0.00005
)

improvement_epochs = 3

best_candidate_val_accuracy = 0.0

improve_train_accuracies = []
improve_val_accuracies = []

save_path = (
    "/content/drive_new/MyDrive/"
    "Food_Recognition_Project/model/"
    "best_improved_mobilenetv2.pth"
)

for epoch in range(improvement_epochs):

    # -------------------------
    # TRAINING
    # -------------------------
    model.train()

    train_correct = 0
    train_total = 0
    train_running_loss = 0.0

    for images, labels in updated_train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        train_total += labels.size(0)
        train_correct += (predicted == labels).sum().item()

    train_loss = (
        train_running_loss / len(updated_train_loader)
    )

    train_accuracy = (
        100 * train_correct / train_total
    )

    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    val_correct = 0
    val_total = 0
    val_running_loss = 0.0

    with torch.no_grad():

        for images, labels in validation_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            val_total += labels.size(0)
            val_correct += (
                predicted == labels
            ).sum().item()

    val_loss = (
        val_running_loss / len(validation_loader)
    )

    val_accuracy = (
        100 * val_correct / val_total
    )

    improve_train_accuracies.append(train_accuracy)
    improve_val_accuracies.append(val_accuracy)

    print(f"Improvement Epoch {epoch+1}/{improvement_epochs}")

    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.2f}%"
    )

    print(
        f"Validation Loss: {val_loss:.4f} | "
        f"Validation Accuracy: {val_accuracy:.2f}%"
    )

    # Save best candidate
    if val_accuracy > best_candidate_val_accuracy:

        best_candidate_val_accuracy = val_accuracy

        torch.save(
            model.state_dict(),
            save_path
        )

        print("✅ Best improvement model saved!")

    print("-" * 50)

print("\nImprovement training finished!")

print(
    f"Best Candidate Validation Accuracy: "
    f"{best_candidate_val_accuracy:.2f}%"
)

print("Previous Best Validation Accuracy: 91.61%")
```

```python
import os
import torch
import pandas as pd
from PIL import Image
from torchvision import transforms, models
import torch.nn as nn

# Paths
test_images_dir = (
    "/content/drive_new/MyDrive/"
    "Food_Recognition_Project/test_images"
)

model_path = (
    "/content/drive_new/MyDrive/"
    "Food_Recognition_Project/model/"
    "best_improved_mobilenetv2.pth"
)

nutrition_path = (
    "/content/drive_new/MyDrive/"
    "Food_Recognition_Project/nutrition/"
    "food_nutrition.csv"
)

# Classes
class_names = [
    'almond', 'apple', 'banana', 'broccoli', 'carrot',
    'cashew', 'chocolate_cake', 'cucumber', 'cupcake',
    'donut', 'french_fries', 'hamburger', 'hot_dog',
    'ice_cream', 'orange', 'peanut', 'pizza',
    'strawberry', 'tomato', 'walnut'
]

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Create and load improved model
model = models.mobilenet_v2(weights=None)

model.classifier[1] = nn.Linear(
    model.last_channel,
    20
)

model.load_state_dict(
    torch.load(model_path, map_location=device)
)

model = model.to(device)
model.eval()

# Image preprocessing
prediction_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Nutrition database
nutrition_df = pd.read_csv(nutrition_path)
nutrition_dict = nutrition_df.set_index("food").to_dict("index")

results = []

valid_extensions = (
    ".jpg", ".jpeg", ".png",
    ".webp", ".jfif"
)

for filename in sorted(os.listdir(test_images_dir)):

    if not filename.lower().endswith(valid_extensions):
        continue

    image_path = os.path.join(
        test_images_dir,
        filename
    )

    image = Image.open(image_path).convert("RGB")

    image_tensor = prediction_transform(image)
    image_tensor = image_tensor.unsqueeze(0).to(device)

    with torch.no_grad():

        output = model(image_tensor)

        probabilities = torch.softmax(
            output,
            dim=1
        )

        confidence, predicted_label = torch.max(
            probabilities,
            1
        )

    predicted_food = class_names[
        predicted_label.item()
    ]

    confidence_percent = confidence.item() * 100

    # Actual food from filename
    actual_food = os.path.splitext(filename)[0]
    actual_food = actual_food.replace("_test", "")

    # Confidence threshold
    if confidence_percent < 50:
        display_result = "Uncertain"
        category = "-"
        status = "Uncertain"

    else:
        display_result = predicted_food.replace(
            "_", " "
        ).title()

        category = nutrition_dict[
            predicted_food
        ]["category"]

        if predicted_food == actual_food:
            status = "Correct"
        else:
            status = "Wrong"

    results.append([
        filename,
        actual_food.replace("_", " ").title(),
        display_result,
        confidence_percent,
        category,
        status
    ])

# Create table
real_world_df = pd.DataFrame(
    results,
    columns=[
        "Image",
        "Actual Food",
        "Model Result",
        "Confidence (%)",
        "Category",
        "Correct?"
    ]
)

real_world_df["Confidence (%)"] = (
    real_world_df["Confidence (%)"].round(2)
)

display(real_world_df)

# Count results
correct_count = (
    real_world_df["Correct?"] == "Correct"
).sum()

uncertain_count = (
    real_world_df["Correct?"] == "Uncertain"
).sum()

wrong_count = (
    real_world_df["Correct?"] == "Wrong"
).sum()

total_count = len(real_world_df)

print("\nCorrect:", correct_count)
print("Uncertain:", uncertain_count)
print("Wrong:", wrong_count)
print("Total:", total_count)

print("\nPrevious Real-World Accuracy: 45.45%")
```

```python
import os
import torch
import torch.nn as nn
import pandas as pd
import matplotlib.pyplot as plt

from torchvision import models, transforms, datasets
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# Paths
project_path = "/content/drive/MyDrive/Food_Recognition_Project"

model_path = os.path.join(
    project_path,
    "model",
    "best_improved_mobilenetv2.pth"
)

test_path = os.path.join(
    project_path,
    "split_data",
    "test"
)

results_dir = os.path.join(
    project_path,
    "results"
)

os.makedirs(results_dir, exist_ok=True)

# Device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Classes
class_names = [
    'almond', 'apple', 'banana', 'broccoli', 'carrot',
    'cashew', 'chocolate_cake', 'cucumber', 'cupcake',
    'donut', 'french_fries', 'hamburger', 'hot_dog',
    'ice_cream', 'orange', 'peanut', 'pizza',
    'strawberry', 'tomato', 'walnut'
]

# Image transform
prediction_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Test dataset
test_dataset = datasets.ImageFolder(
    test_path,
    transform=prediction_transform
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

# Load improved model
model = models.mobilenet_v2(weights=None)

model.classifier[1] = nn.Linear(
    model.last_channel,
    20
)

model.load_state_dict(
    torch.load(model_path, map_location=device)
)

model = model.to(device)
model.eval()

# Predict test dataset
all_labels = []
all_predictions = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        all_labels.extend(labels.numpy())
        all_predictions.extend(
            predicted.cpu().numpy()
        )

# Final accuracy
correct = sum(
    p == y
    for p, y in zip(
        all_predictions,
        all_labels
    )
)

final_accuracy = (
    correct / len(all_labels) * 100
)

print("Final Test Images:", len(all_labels))
print("Correct Predictions:", correct)
print(f"Final Test Accuracy: {final_accuracy:.2f}%")
```

```python
# ==========================================
# Save final classification report
# ==========================================

report = classification_report(
    all_labels,
    all_predictions,
    target_names=class_names,
    digits=4,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()

report_df.to_csv(
    os.path.join(
        results_dir,
        "final_classification_report.csv"
    )
)

print("Final classification report updated!")


# ==========================================
# Save final confusion matrix
# ==========================================

cm = confusion_matrix(
    all_labels,
    all_predictions
)

fig, ax = plt.subplots(figsize=(14, 14))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    ax=ax,
    xticks_rotation=90
)

plt.title("Final Improved MobileNetV2 Confusion Matrix")
plt.tight_layout()

plt.savefig(
    os.path.join(
        results_dir,
        "final_confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Final confusion matrix updated!")


# ==========================================
# Update model comparison
# ==========================================

comparison_df = pd.DataFrame({
    "Model": [
        "Original MobileNetV2",
        "Fine-Tuned MobileNetV2",
        "Improved MobileNetV2"
    ],
    "Test Accuracy (%)": [
        92.28,
        95.18,
        95.50
    ]
})

comparison_df.to_csv(
    os.path.join(
        results_dir,
        "model_comparison.csv"
    ),
    index=False
)

display(comparison_df)

print("Model comparison updated!")
```

```python
# ==========================================
# FINAL FOOD IMAGE TESTING
# ==========================================

import os
import torch
import torch.nn as nn
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from torchvision import models, transforms
from google.colab import files


# ==========================================
# 1. DEVICE
# ==========================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)


# ==========================================
# 2. FOOD CLASSES
# ==========================================

class_names = [
    'almond',
    'apple',
    'banana',
    'broccoli',
    'carrot',
    'cashew',
    'chocolate_cake',
    'cucumber',
    'cupcake',
    'donut',
    'french_fries',
    'hamburger',
    'hot_dog',
    'ice_cream',
    'orange',
    'peanut',
    'pizza',
    'strawberry',
    'tomato',
    'walnut'
]


# ==========================================
# 3. LOAD FINAL IMPROVED MODEL
# ==========================================

model_path = (
    "/content/drive/MyDrive/"
    "Food_Recognition_Project/model/"
    "best_improved_mobilenetv2.pth"
)

model = models.mobilenet_v2(weights=None)

model.classifier[1] = nn.Linear(
    model.last_channel,
    20
)

model.load_state_dict(
    torch.load(
        model_path,
        map_location=device
    )
)

model = model.to(device)
model.eval()

print("Final improved model loaded!")


# ==========================================
# 4. IMAGE PREPROCESSING
# ==========================================

prediction_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ==========================================
# 5. LOAD NUTRITION DATABASE
# ==========================================

nutrition_path = (
    "/content/drive/MyDrive/"
    "Food_Recognition_Project/nutrition/"
    "food_nutrition.csv"
)

nutrition_df = pd.read_csv(
    nutrition_path
)

nutrition_dict = (
    nutrition_df
    .set_index("food")
    .to_dict("index")
)

print("Nutrition database loaded!")


# ==========================================
# 6. CHOOSE NUMBER OF IMAGES
# ==========================================

number_of_images = int(
    input("\nHow many food images do you want to test? ")
)

if number_of_images <= 0:
    raise ValueError(
        "Please enter a number greater than 0."
    )

print(
    f"\nPlease upload exactly "
    f"{number_of_images} food image(s)."
)


# ==========================================
# 7. UPLOAD IMAGES
# ==========================================

uploaded = files.upload()

uploaded_files = list(
    uploaded.keys()
)

# Check number of uploaded files
if len(uploaded_files) != number_of_images:

    print(
        f"\nYou selected {number_of_images} image(s), "
        f"but uploaded {len(uploaded_files)} file(s)."
    )

    print(
        "Please run this cell again and upload "
        "the correct number of images."
    )

    raise ValueError(
        "Number of uploaded images does not match."
    )

print("\nUploaded files:")

for filename in uploaded_files:
    print("-", filename)


# ==========================================
# 8. ENTER WEIGHT FOR EACH IMAGE
# ==========================================

weights = {}

print("\nEnter estimated weight for each food:")

for filename in uploaded_files:

    while True:

        try:

            portion = float(
                input(
                    f"Weight for {filename} (grams): "
                )
            )

            if portion <= 0:
                print(
                    "Weight must be greater than 0."
                )
                continue

            weights[filename] = portion
            break

        except ValueError:

            print(
                "Please enter a valid number."
            )

print("\nAll weights entered.")
print("Starting food prediction...")


# ==========================================
# 9. PREDICT ALL IMAGES
# ==========================================

demo_results = []

for filename in uploaded_files:

    portion_grams = weights[filename]

    # Open image
    original_image = Image.open(
        filename
    ).convert("RGB")

    # Prepare image
    image_tensor = prediction_transform(
        original_image
    )

    image_tensor = (
        image_tensor
        .unsqueeze(0)
        .to(device)
    )

    # Prediction
    with torch.no_grad():

        output = model(
            image_tensor
        )

        probabilities = torch.softmax(
            output,
            dim=1
        )

        confidence, predicted_label = torch.max(
            probabilities,
            1
        )

    predicted_food = class_names[
        predicted_label.item()
    ]

    confidence_percent = (
        confidence.item() * 100
    )


    # ======================================
    # LOW CONFIDENCE
    # ======================================

    if confidence_percent < 50:

        result_name = "Uncertain Prediction"
        category = "-"

        calories = "-"
        protein = "-"
        carbs = "-"
        fat = "-"


    # ======================================
    # CONFIDENT PREDICTION
    # ======================================

    else:

        result_name = (
            predicted_food
            .replace("_", " ")
            .title()
        )

        nutrition = nutrition_dict[
            predicted_food
        ]

        category = nutrition[
            "category"
        ]

        ratio = (
            portion_grams / 100
        )

        calories = round(
            nutrition["calories_kcal"]
            * ratio,
            2
        )

        protein = round(
            nutrition["protein_g"]
            * ratio,
            2
        )

        carbs = round(
            nutrition["carbs_g"]
            * ratio,
            2
        )

        fat = round(
            nutrition["fat_g"]
            * ratio,
            2
        )


    # Save result
    demo_results.append([
        filename,
        result_name,
        round(
            confidence_percent,
            2
        ),
        category,
        portion_grams,
        calories,
        protein,
        carbs,
        fat
    ])


# ==========================================
# 10. SHOW RESULTS ONE BY ONE
# ==========================================

print("\n")
print("=" * 50)
print("FOOD RECOGNITION RESULTS")
print("=" * 50)

for i, result in enumerate(
    demo_results
):

    filename = result[0]
    predicted_food = result[1]
    confidence = result[2]
    category = result[3]
    portion = result[4]
    calories = result[5]
    protein = result[6]
    carbs = result[7]
    fat = result[8]

    print("\n")
    print("=" * 50)
    print(
        f"RESULT {i + 1} "
        f"OF {len(demo_results)}"
    )
    print("=" * 50)

    print(
        "Image:",
        filename
    )

    # Uncertain
    if predicted_food == "Uncertain Prediction":

        print(
            "Result: Uncertain Prediction"
        )

        print(
            f"Confidence: {confidence}%"
        )

        print(
            "Category: -"
        )

        print("\nNutrition:")
        print(
            "Not shown because confidence "
            "is below 50%."
        )

    # Confident
    else:

        print(
            "Predicted Food:",
            predicted_food
        )

        print(
            f"Confidence: {confidence}%"
        )

        print(
            "Category:",
            category
        )

        print("\nEstimated Nutrition")
        print("-------------------------")

        print(
            f"Portion Size: {portion} g"
        )

        print(
            f"Calories: {calories} kcal"
        )

        print(
            f"Protein: {protein} g"
        )

        print(
            f"Carbohydrates: {carbs} g"
        )

        print(
            f"Fat: {fat} g"
        )


    # ======================================
    # SHOW IMAGE
    # ======================================

    original_image = Image.open(
        filename
    ).convert("RGB")

    plt.figure(
        figsize=(5, 5)
    )

    plt.imshow(
        original_image
    )

    if predicted_food == "Uncertain Prediction":

        plt.title(
            f"Uncertain Prediction\n"
            f"Confidence: {confidence}%"
        )

    else:

        plt.title(
            f"{predicted_food}\n"
            f"Confidence: {confidence}%"
        )

    plt.axis("off")
    plt.show()


# ==========================================
# 11. SAVE RESULTS TO GOOGLE DRIVE
# ==========================================

demo_df = pd.DataFrame(
    demo_results,
    columns=[
        "Image",
        "Predicted Food",
        "Confidence (%)",
        "Category",
        "Portion (g)",
        "Calories (kcal)",
        "Protein (g)",
        "Carbohydrates (g)",
        "Fat (g)"
    ]
)

results_dir = (
    "/content/drive/MyDrive/"
    "Food_Recognition_Project/results"
)

os.makedirs(
    results_dir,
    exist_ok=True
)

demo_result_path = os.path.join(
    results_dir,
    "final_live_demo_results.csv"
)

demo_df.to_csv(
    demo_result_path,
    index=False
)

print("\n" + "=" * 50)
print("TESTING COMPLETED!")
print("=" * 50)

print(
    "Total images tested:",
    len(demo_results)
)

print("\nResults saved to:")
print(
    demo_result_path
)
```
