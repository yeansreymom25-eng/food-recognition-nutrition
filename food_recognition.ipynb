```python
import os
import requests

# Fruits-360 Almond folder
url = "https://api.github.com/repos/fruits-360/fruits-360-100x100/contents/Training/Almonds%201"

# Your Google Drive almond folder
save_folder = "/content/drive/MyDrive/Food_Recognition_Project/dataset/almond"

os.makedirs(save_folder, exist_ok=True)

files = requests.get(url).json()

images = [
    file for file in files
    if file["name"].lower().endswith((".jpg", ".jpeg", ".png"))
]

# Use only 150 images
images = images[:150]

for number, image in enumerate(images, start=1):
    image_data = requests.get(image["download_url"]).content

    save_path = os.path.join(save_folder, image["name"])

    with open(save_path, "wb") as f:
        f.write(image_data)

    print(f"{number}/{len(images)} downloaded")

print("Almond finished!")
```

```python
import os
import requests
import zipfile
import glob
import shutil

# Download broccoli dataset
url = "https://ghaipublic.s3.us-west-2.amazonaws.com/datasets/broccoli_annotated.zip"

zip_path = "/content/broccoli.zip"
extract_folder = "/content/broccoli_data"

print("Downloading broccoli dataset...")

r = requests.get(url)
with open(zip_path, "wb") as f:
    f.write(r.content)

print("Download finished!")

# Extract
os.makedirs(extract_folder, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_folder)

print("Extract finished!")

# Find all images
images = glob.glob(extract_folder + "/**/*.jpg", recursive=True)
images += glob.glob(extract_folder + "/**/*.jpeg", recursive=True)
images += glob.glob(extract_folder + "/**/*.png", recursive=True)

print("Images found:", len(images))

# Google Drive broccoli folder
save_folder = "/content/gdrive/MyDrive/Food_Recognition_Project/dataset/broccoli"
os.makedirs(save_folder, exist_ok=True)

# Take only 150
images = images[:150]

for number, image in enumerate(images, start=1):

    new_name = f"broccoli_{number}.jpg"

    shutil.copy2(
        image,
        os.path.join(save_folder, new_name)
    )

print("Broccoli finished!")
print("Images saved:", len(os.listdir(save_folder)))
```

```python
import os
from PIL import Image

dataset_path = "/content/gdrive/MyDrive/Food_Recognition_Project/dataset"

bad_images = []

for folder in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder)

    if not os.path.isdir(folder_path):
        continue

    for file in os.listdir(folder_path):

        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            file_path = os.path.join(folder_path, file)

            try:
                img = Image.open(file_path)
                img.verify()
            except:
                bad_images.append(file_path)

print("Broken images found:", len(bad_images))

for image in bad_images[:20]:
    print(image)
```
